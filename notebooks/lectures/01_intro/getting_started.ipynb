# Getting Started

This notebook verifies your environment is set up correctly and shows how to access course data. 

All course data lives in a single file, pointed to by the link you see below. The cell below downloads that file and attaches it for usage with duckdb, it might take up to a minute (depending on your internet speed).

**If this notebook runs successfully, your setup is complete. If it does not (you see errors when running cells), please let me know ASAP.**

```python
import duckdb
import gzip
import urllib.request
from pathlib import Path

data_url = "https://pub-c88b3a7f2ed141418355b2bfb03c96e6.r2.dev/data.duckdb.gz"
db_path = "data.duckdb"

request = urllib.request.Request(data_url, headers={"User-Agent": "Mozilla/5.0"})
with urllib.request.urlopen(request) as response:
    with gzip.open(response, "rb") as gz:
        with open(db_path, "wb") as f:
            f.write(gz.read())

conn = duckdb.connect()
conn.execute(f"ATTACH '{db_path}' AS data (READ_ONLY)")
```

```python
conn.execute("SHOW TABLES FROM data.attendance").df()
```

```python
conn.execute("SELECT * FROM data.attendance.attendance LIMIT 10").df()
```
