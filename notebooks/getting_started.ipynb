# Getting Started

This notebook verifies your environment is set up correctly and shows how to access course data. 

All course data lives in a single file, pointed to by the link you see below. The cell below downloads that file and attaches it for usage with duckdb.

```python
import duckdb
import urllib.request
from pathlib import Path

data_url = "https://pub-c88b3a7f2ed141418355b2bfb03c96e6.r2.dev/data.duckdb"
db_path = Path("course_data.duckdb")

if not db_path.exists():
    print("Downloading database...")
    request = urllib.request.Request(data_url, headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(request) as response, open(db_path, "wb") as f:
        f.write(response.read())

conn = duckdb.connect()
conn.execute(f"ATTACH '{db_path}' AS course_data (READ_ONLY)")

print("Connected successfully!")
```

```python
conn.execute("SHOW TABLES FROM course_data.admissions").df()
```

```python
conn.execute("SELECT * FROM course_data.admissions.applications LIMIT 10").df()
```
