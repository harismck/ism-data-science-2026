```python
import pandas as pd
import duckdb
import seaborn as sns

db_path = "data.duckdb"
schema = "data.vehicle_inspection"

conn = duckdb.connect()
conn.execute(f"ATTACH '{db_path}' AS data (READ_ONLY)")
conn.execute(f"USE {schema}");
```

```python
df = conn.sql("""
    SELECT
    * -- select just the necessary columns
    FROM inspection
    WHERE class='M1' 
    AND inspection_type_en = 'Technical inspection'
    AND inspection_date_week > '2019-05-01'
""").to_df()

# If you find that the notebook works very slowly or crashes, downsample

# df = conn.sql("""
#     SELECT
#     * -- select just the necessary columns
#     FROM inspection
#     WHERE class='M1' 
#     AND inspection_type_en = 'Technical inspection'
#     AND inspection_date_week > '2019-05-01'
#     ORDER BY random()
#     LIMIT 1000000
# """).to_df()
```
